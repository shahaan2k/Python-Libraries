# Students Performance DataSet: 

```python
import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Load The DataSet
data = pd.read_csv("Student_Performance_Dataset_1000.csv")
data

# X input Data Y Output Data
x = data[['Study_Hours_per_Week']]
y = data['Final_Score'] 

# Train Model 
model = LinearRegression()

model.fit(x,y) 

predicted_score = model.predict(x) 

# Valid Regression Metrics 
mae = mean_absolute_error(y,predicted_score)
mse = mean_squared_error(y,predicted_score)
rmse = np.sqrt(mse)
r2 = r2_score(y, predicted_score)

# Printing The Results 
print("Mean Absolute Error: ", round(mae,2)) # round function shows 2 digits after decimal point 
print("Mean Squared Error: ", round(mse,2))
print("Root Mean Squared Error: ", round(rmse,2)) 
print("R^2 Score (Model Accuracy): ", round(r2,4)) # If closer to 1 = better model 

# Histogram 
plt.figure(figsize=(10,6))
plt.hist(data['Final_Score'], bins=30, color='skyblue', edgecolor='black')
plt.title("Distribution Of Final Exam Score")
plt.xlabel("Final Exam Score")
plt.ylabel("Number Of Students")
plt.grid(True)
plt.show()

# Scatter Plot And Regression Line 
plt.figure(figsize=(10,6))
plt.scatter(x,y , color='green', label='Actual Score')
plt.plot(x, predicted_score, color='red', label='Predicted Score')
plt.title("Model Prediction VS Actual Score")
plt.xlabel("Study Hours Per Week")
plt.ylabel("Final Marks")
plt.grid(True)
plt.show()

# Predicting A New Score By Number Of Hours
new_hours = 9
pred_new_score = model.predict([[new_hours]])
print(f"Predicted Final Score For {new_hours} Hours Is {pred_new_score} Marks")      
      
```
