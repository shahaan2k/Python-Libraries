# Elbow Method For K-Means 

```python
import pandas as pd 
from sklearn.cluster import KMeans 
import matplotlib.pyplot as plt 
```

```python
# Sample Data 
data = {
    'Customer': ['Shahan','Dayyan','Khadim','Sudais','Arsalan','Hammad'],
    'Age':[20,18,24,19,20,20],
    'Spending':[100,200,300,110,290,130]
}

df = pd.DataFrame(data)

x = df[['Age','Spending']]

model = KMeans(n_clusters=2, random_state=42, n_init=10)

df['Group'] = model.fit_predict(x) 

plt.figure(figsize=(6,5))
for group in df['Group'].unique():
    group_data = df[df['Group']== group]
    plt.scatter(group_data['Age'], group_data['Spending'], label =f'Group {group}')

plt.xlabel('Age')    
plt.ylabel('Spending Scores')
plt.title('Customer Segments (K-Means)')
plt.legend()
plt.grid(True)
plt.show() 

print(df) 

# Note: This program shows how the k means groups the data so that the learner can see how it works 
```

# Elbow Graph:

```python
# Sample Data 
data = {
    'Customer': ['Shahan','Dayyan','Khadim','Sudais','Arsalan','Hammad'],
    'Age':[20,18,24,19,20,20],
    'Spending':[100,200,300,110,290,130]
}

df = pd.DataFrame(data)

# Select Features
X = df[['Age', 'Spending']]

# -----------------------------
# Elbow Method
# -----------------------------

wcss = []

for k in range(1, 6):   # Try K values from 1 to 5
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X)
    wcss.append(model.inertia_)   # Store WCSS

# Plot Elbow Graph
plt.figure(figsize=(6,5))
plt.plot(range(1,6), wcss, marker='o')
plt.title("Elbow Method")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.grid(True)
plt.show()
```
