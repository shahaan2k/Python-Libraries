```python
import cv2
import numpy as np
```

```python
# Canny Function 

img = cv2.imread("flower.png", cv2.IMREAD_GRAYSCALE)

edges = cv2.Canny(img, 50, 150)

cv2.imshow("Orignal Image:", img)
cv2.imshow("Canny Image:", edges)
cv2.waitKey(0)
cv2.destroyAllWindows() 
```

```python
# Threshold Function: 

img = cv2.imread("flower.png", cv2.IMREAD_GRAYSCALE)

ret, thresh_img = cv2.threshold(img, 44, 255, cv2.THRESH_BINARY)

cv2.imshow("Orignal Image:", img)
cv2.imshow("Threshold Image:", thresh_img)
cv2.waitKey(0)
cv2.destroyAllWindows() 

""""
90 - 0 Black
130-255 white
180-255 white
50-0 black
```

```python
# Bitwise Operations: 

img1 = np.zeros((300,300), dtype="uint8")
img2 = np.zeros((300,300), dtype="uint8")

cv2.circle(img1, (150,150), 100, 255, -1)
cv2.rectangle(img2, (150,150), (250,250), 255, -1)

bitwise_and = cv2.bitwise_and(img1,img2)
bitwise_or = cv2.bitwise_or(img1,img2)
bitwise_not = cv2.bitwise_not(img1)

cv2.imshow("Circle", img1)
cv2.imshow("Rectangle", img2)
cv2.imshow("And", bitwise_and)
cv2.imshow("Or", bitwise_or)
cv2.imshow("Not", bitwise_not)
cv2.waitKey(0)
cv2.destroyAllWindows() 
```

```python
# Countour Functions: 

img = cv2.imread("hexagon.png")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

_, thresh = cv2.threshold(gray, 200,220,cv2.THRESH_BINARY)

# Finding Contours 
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

cv2.drawContours(img, contours, -1, (0,255,0), 3)

cv2.imshow("COntours", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Shape Detection:

img = cv2.imread("hexagon.png")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

_, thresh = cv2.threshold(gray, 200,255,cv2.THRESH_BINARY)

# Finding Contours 
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

cv2.drawContours(img, contours, -1, (0,255,0), 3)

for contour in contours:
    approx = cv2.approxPolyDP(contour, 0.01 * cv2.arcLength(contour,True), True)

    corners = len(approx)

    if corners ==3:
        shape = "Triangle"
    elif corners ==4:
        shape = "Square OR Rectangle"
    elif corners ==5:
        shape = "Pentagon"
    elif corners ==6:
        shape = "Hexagon"
    elif corners >6:
        shape = "Circle"
    else:
        print("Unknown")

    cv2.drawContours(img,[approx], -1, (0, 255, 0), 2)
    x = approx.ravel()[0]
    y = approx.ravel()[1]
    cv2.putText(img, shape, (x,y), cv2.FONT_HERSHEY_COMPLEX, 0.6, (255, 0, 0), 2)
        

cv2.imshow("COntours", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
