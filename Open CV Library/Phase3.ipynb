```python
import cv2
```

```python
# Image Drawing Functions:
```

```python
# Drawing A Line

image = cv2.imread("loc.png")

if image is None:
    print("Image Not Found")
else:
    print("Imaged Loaded Successfully")

    pt1 = (50,100) # From X axis To Y axis
    pt2 = (300,100) # From X axis To Y axis
    color = (255,0,0) 
    thickness = 4 # Line Thickness

    cv2.line(image, pt1, pt2, color, thickness) # Line Drwaing Function 

    cv2.imshow("Line Drawing", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows() 
```

```python
# Drawimg A Rectangle:

image = cv2.imread("loc.png")

if image is None:
    print("Image Not Found")
else:
    print("Imaged Loaded Successfully")

    pt1 = (50,100) # From X axis To Y axis
    pt2 = (250,200) # From X axis To Y axis
    color = (255,0,0) 
    thickness = 4 # Line Thickness

    cv2.rectangle(image, pt1, pt2, color, thickness) # Line Drwaing Function 

    cv2.imshow("Image Focusing Rectangle", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows() 
```

```python
# Drawing A Circle On An Image:

image = cv2.imread("loc.png")

if image is None:
    print("Image Not Found")
else:
    print("Imaged Loaded Successfully")

    cv2.circle(image, (150,150) , 50, (255,100,200), -1) # cv2.circle(image, center, radius, color, thickness) 

    cv2.imshow("Circle On An Image:", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows() 
```

```python
# Adding Text On Image:

image = cv2.imread("loc.png")

if image is None:
    print("Image Not Found")
else:
    print("Imaged Loaded Successfully")

    cv2.putText(                # cv2.putText(img, text, org, font, fontScale, color, thickness)
        image,
        "This Too Shall Pass",
        (50,400),
        cv2.FONT_HERSHEY_SIMPLEX,
        1.2,
        (255,0,0),
        2
    )

    cv2.imshow("Adding Text On An Image:", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows() 
```

```python
# Video Capturing:

cap = cv2.VideoCapture(0) # for external webcam=1 , for internal webcam=0

while True:
    ret, frame = cap.read() # ret= True/False weather image is reading or not And frame=image

    if not ret:
        print("Could Not Read Frame")
        break
    cv2.imshow("Webcam Feed", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        print("Quitting...")
        break
cap.release()
cv2.destroyAllWindows()
```

```python
# Saving Video:

camera = cv2.VideoCapture(0)

frame_width = int(camera.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(camera.get(cv2.CAP_PROP_FRAME_HEIGHT))

codec = cv2.VideoWriter_fourcc(*'XVID')
recorder = cv2.VideoWriter("My Video.avi", codec, 20, (frame_width, frame_height))

while True:
    success, image = camera.read()

    if not success:
        break

    recorder.write(image)
    cv2.imshow("Recording Live", image)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

camera.release()
recorder.release()
```
