```python
import cv2
import numpy as np
```

```python
# Image Blurring:

# Gaussian Blur: 

image = cv2.imread("nature.png")

blurred = cv2.GaussianBlur(image, (7,7), 3)

cv2.imshow("Orignal Image", image)
cv2.imshow("Blurred Image", blurred)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Median Blur:

image = cv2.imread("nature.png")

blurred = cv2.medianBlur(image, 21)

cv2.imshow("Orignal Image", image)
cv2.imshow("Median Blur Image", blurred)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Image Sharpening:

image = cv2.imread("lowr.png")

sharpen = np.array([
    [0,-1,0],
    [-1,5,-1],
    [0,-1,0],
])

sharpened = cv2.filter2D(image, -1, sharpen)

cv2.imshow("Orignal Image", image)
cv2.imshow("Sharpened Image", sharpened)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
