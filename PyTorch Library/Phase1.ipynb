## PyTorch:

```python
import torch 
```

```python
x = torch.tensor([1,2,3,4,5]) # 1D Tensor
print(x)
```

```python
y = torch.tensor([ # 2D Tensor 
    [1,2],
    [3,4]    
])

print(y)
```

```python
# Zeros Array
z = torch.zeros(3,2) # 3 Rows 2 Columns
print(z)
```

```python
# Ones Array
o = torch.ones(3,2) # 3 Rows 2 Columns
print(o)
```

```python
# Random Decimal Array
r = torch.rand(2,2)
print(r)
```

```python
# Tensor Shapes 
s = torch.tensor([
    [1,2,3],
    [4,5,6]
])
print(s.shape)
```

```python
# Indexing 
i = torch.tensor([10,20,30])
print(f'One Dimensional Indexing: \n{i[0]}')

# 2D Indexing
d = torch.tensor([
    [1,2,3],
    [4,5,6]
])
print(f'Two Dimensional Indexing: \n {d[1,2]}')
```

# Basic Functions:

```python
# Addiction & Multiplication
a = torch.tensor([1,2,3])
b = torch.tensor([4,5,6])

print(a+b)
print(a*b)
```
